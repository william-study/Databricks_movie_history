```python
movie_genre_name_df = movie_genre_id_df \
                           .join(genre_df,
                                 movie_genre_id_df.genre_Id == genre_df.genre_Id, 
                                 "inner") \
                           .select(movie_genre_id_df.budget,
                                   movie_genre_id_df.revenue,
                                   movie_genre_id_df.year_Release_Date,
                                   genre_df.genre_Name)
```

```python
display(movie_genre_name_df)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import sum, rank, dense_rank, asc, desc, lit
```

```python
movie_rank = Window.partitionBy("year_Release_Date").orderBy(desc("total_budget"), desc("total_revenue"))
movie_dense_rank = Window.partitionBy("year_Release_Date").orderBy(desc("total_budget"), desc("total_revenue"))

movie_genre_total_final_df = movie_genre_name_df \
                         .filter("year_Release_Date >= 2015") \
                         .groupBy("year_Release_Date", "genre_Name") \
                             .agg(
                                 sum("budget").alias("total_budget"),
                                 sum("revenue").alias("total_revenue")
                                 ) \
                          .withColumn("rank", rank().over(movie_rank)) \
                          .withColumn("dense_rank", dense_rank().over(movie_dense_rank)) \
                          .withColumn("created_date", lit(v_file_date))
```

```python
display(movie_genre_total_final_df)
```
