```sql
/*El

size --> me devuelve la cantidad elementos del arreglo
'\\s+' --> significa si encuentra dos o mas valores en blanco entre los caracteres
get --> lo que encuentre lo devolvera a la posicion que indica despues de la "," 
concat_ws --> concatenara cada palabra que encuentre separadas por ' ' 
coalesce --> si encuentra valor, devuelve lo que encuentra, de lo contrario devuelve '' */

SELECT *,
       CASE
         WHEN size(split(trim(name), '\\s+')) >= 3 THEN concat_ws(' ', get(split(trim(name), '\\s+'), 0), get(split(trim(name), '\\s+'), 1))
         ELSE coalesce(get(split(trim(name), '\\s+'), 0), '')
       END AS forename,
       CASE
         WHEN size(split(trim(name), '\\s+')) >= 3 THEN coalesce(get(split(trim(name), '\\s+'), 2), '')
         ELSE coalesce(get(split(trim(name), '\\s+'), 1), '')
       END AS surname
FROM persons;
```

####Agreggate Functions (Funciones de Agregacion)

```sql
SELECT COUNT(1)
FROM movies;
```

```sql
SELECT max(release_date)
FROM movies;
```

```sql
SELECT *
FROM movies
WHERE year_Release_Date = 2010;
```

```sql
SELECT year_Release_Date, COUNT(1) AS total_count, SUM(budget) sum_budget,
MAX(budget) max_budget, MIN(budget) min_budget, AVG(budget) avg_budget
FROM movies
GROUP BY year_Release_Date
HAVING COUNT(1) > 220
ORDER BY year_Release_Date;
```

####Window Functions (Funciones de Ventana)

```sql
SELECT title, year_Release_Date, release_Date,
       RANK() OVER(PARTITION BY year_Release_Date ORDER BY release_Date) AS rank
FROM movies
WHERE year_Release_Date IS NOT NULL
ORDER BY year_Release_Date;
```

####Join

```sql
DESCRIBE movies;
```

```sql
SELECT movie_Id, title, budget, revenue, year_Release_Date, release_Date
FROM movies
WHERE year_Release_Date = 2011;
```

```sql
CREATE OR REPLACE TEMP VIEW v_movies_2011
AS
SELECT movie_Id, title, budget, revenue, year_Release_Date, release_Date
FROM movies
WHERE year_Release_Date = 2011;
```

```sql
SELECT * FROM v_movies_2011;
```

```sql
CREATE OR REPLACE TEMP VIEW v_productions_countries
AS
SELECT *
FROM production_country
```

```sql
SELECT * FROM v_productions_countries
```

```sql
CREATE OR REPLACE TEMP VIEW v_countries
AS
SELECT country_Id, country_Name FROM countries
```

```sql
SELECT * FROM v_countries
```

####Inner Join

```sql
SELECT VM.title, VM.budget, VM.revenue, VM.year_Release_Date, VM.release_Date, VC.country_Name
FROM v_movies_2011 VM
INNER JOIN v_productions_countries VPC ON VM.movie_Id = VPC.movie_Id
INNER JOIN v_countries VC ON VC.country_Id = VPC.country_Id;
```

####Left Join

```sql
SELECT VM.title, VM.budget, VM.revenue, VM.year_Release_Date, VM.release_Date, VC.country_Name
FROM v_movies_2011 VM
LEFT JOIN v_productions_countries VPC ON VM.movie_Id = VPC.movie_Id
LEFT JOIN v_countries VC ON VC.country_Id = VPC.country_Id;
```

####Right Join

```sql
SELECT VM.title, VM.budget, VM.revenue, VM.year_Release_Date, VM.release_Date, VC.country_Name
FROM v_movies_2011 VM
RIGHT JOIN v_productions_countries VPC ON VM.movie_Id = VPC.movie_Id
RIGHT JOIN v_countries VC ON VC.country_Id = VPC.country_Id;
```

####Full Join

```sql
SELECT VM.title, VM.budget, VM.revenue, VM.year_Release_Date, VM.release_Date, VC.country_Name
FROM v_movies_2011 VM
FULL JOIN v_productions_countries VPC ON VM.movie_Id = VPC.movie_Id
FULL JOIN v_countries VC ON VC.country_Id = VPC.country_Id;
```

####Semi Join

```sql
SELECT VM.title, VM.budget, VM.revenue, VM.year_Release_Date, VM.release_Date
FROM v_movies_2011 VM
INNER JOIN v_productions_countries VPC ON VM.movie_Id = VPC.movie_Id
SEMI JOIN v_countries VC ON VC.country_Id = VPC.country_Id;
```

####Anti Join

```sql
SELECT VM.title, VM.budget, VM.revenue, VM.year_Release_Date, VM.release_Date
FROM v_movies_2011 VM
INNER JOIN v_productions_countries VPC ON VM.movie_Id = VPC.movie_Id
ANTI JOIN v_countries VC ON VC.country_Id = VPC.country_Id;
```

####Cross Join

```sql
SELECT VM.title, VM.budget, VM.revenue, VM.year_Release_Date, VM.release_Date, VC.country_Name
FROM v_movies_2011 VM CROSS JOIN v_countries VC;
```
