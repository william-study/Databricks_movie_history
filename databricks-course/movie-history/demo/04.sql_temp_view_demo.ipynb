```python
results_movie_genre_language_df = spark.read.parquet(f"{gold_folder_path}/results_movie_genre_language")
```

```python
## Esta vista que se creara, desaparecera si el cluster se apaga o desvincula y solo
## funciona en esta notebook, y para no incurrir en error si se crea una vez con 
## createTempView ya no se podra volver a crear si se quiere corregir un dato
## recomendacion usar createOrReplaceTempView

results_movie_genre_language_df.createOrReplaceTempView("v_movie_genre_language")
```

```python
%sql
SELECT *
FROM v_movie_genre_language
WHERE vote_Average > 7.5
```

```python
results_movie_genre_language_2_df = spark.sql("SELECT * FROM v_movie_genre_language WHERE vote_Average > 7.5")
```

```python
display(results_movie_genre_language_2_df)
```

```python
spark.sql("""
           SELECT * 
           FROM v_movie_genre_language 
           WHERE vote_Average > 7.5
           """) \
     .display()
```

####Global Temp View
1. Crear una vista (view) temporal global del DataFrame
2. Acceder a la vista (view) desde la celda "SQL"
3. Acceder a la vista (view) desde la celda "Python"
4. Acceder a la vista desde otro "Notebook"

```python
## Se creara una tabla global, con el especial detalle que no se crea en la misma base
## de datos que se esta usando, sino en la base de datos global, por lo que para saber
## cual es, podemos usar la sentencia SHOW TABLES IN global_temp, si se usa solo la 
## sentencia SHOW TABLES nos arrojara solo las tablas locales

results_movie_genre_language_df.createOrReplaceGlobalTempView("gv_movie_genre_language")
```

```python
%sql
SHOW TABLES IN global_temp
```

```python
%sql
SELECT *
FROM global_temp.gv_movie_genre_language
```

```python
spark.sql("""
           SELECT * 
           FROM global_temp.gv_movie_genre_language
           """) \
     .display()
```
